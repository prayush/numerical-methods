# PDEs & Numerical Methods

## 1. Introduction: PDEs in the Physical World

**Goal:** Develop a description of the physical world in terms of quantities of interest that can vary across **space** as well as **time**.

**Examples of Quantities:**

* Temperature distribution across a metal plate.

* Propagation of pressure / light waves.

* Electron density in molecules.

* Fluid motion (atmosphere, waterfalls).

* Drop of ink in water.

**Definition:**
These are equations involving unknown functions of multiple variables and their partial derivatives.

$$
u(x, y, z, t)
$$

$\partial_t u(x, y, z, t)$                   $\partial_x u(x, y, z, t)$                 $\partial_y u(x, y, z, t)$                   $\partial_z u(x, y, z, t)$

$\Rightarrow f\left(u, \{\partial_k u\}_{k=t,x,y,z}, \{\partial^2_k u\}, \cdots \{\partial^N_k u\}, \right)=0$

## 2. PDEs in Action

Here are five fundamental examples of PDEs:

### 1. Heat / Diffusion Equation

$$
\frac{\partial u}{\partial t} = \kappa \nabla^2 u
$$

* **Physics & Applications:** $\kappa$ = Diffusivity.

  * **Heat Conduction:** $\kappa$ is *Thermal Diffusivity*. It measures how fast a material reacts to temperature changes (ratio of conducted to stored heat).

  * **Particle Diffusion:** $\kappa$ is the *Diffusion Coefficient*, measuring the rate of spreading from high to low concentration.

  * **Image Processing:** Used for Gaussian blurring to smooth out noise.

  * **Financial Mathematics:** The Black-Scholes equation, used to model the price of options over time, is a transformation of the heat equation. Here, $\kappa$ relates to the **volatility** ($\sigma^2$) of the underlying asset.

  * **Biology:** Models the dispersal of biological populations (e.g., bacteria, invasive species) across a landscape. $\kappa$ represents the **rate of population spread**.

* **Nature:** Parabolic.

* **Parametric Regimes:**

  * **Positive (**$\kappa > 0$**):** Standard diffusion. Disturbances decay and smooth out over time (system evolves to equilibrium).

  * **Negative (**$\kappa < 0$**):** "Anti-diffusion" (ill-posed). Corresponds to running diffusion in reverse; small disturbances grow exponentially (unstable).

  * **Complex (**$\kappa \in \mathbb{C}$**):** If $\kappa$ is imaginary (e.g., $i\hbar/2m$), this becomes the **Schrödinger Equation**. The solution oscillates (waves) rather than decays.

### 2. Wave Equation

$$
\frac{\partial^2 U}{\partial t^2} = c^2 \nabla^2 U
$$

* **Physics & Applications:** $c^2$ = Square of propagation speed.

  * **Acoustics:** Propagation of sound waves in air or fluids. $c$ is the speed of sound.

  * **Electromagnetism:** Propagation of light or radio waves. $c$ is the speed of light.

  * **Elasticity:** Vibrations on a string (1D), membrane (2D), or solid structure. Here, $c$ is determined by the ratio of stiffness/tension to density ($c = \sqrt{T/\mu}$).

  * **Fluid Dynamics:** Shallow water waves.

* **Characteristics:** Finite-speed propagation of information/energy.

* **Nature:** Hyperbolic.

* **Parametric Regimes:**

  * **Positive (**$c^2 > 0$**):** Standard hyperbolic wave equation. Solutions oscillate and propagate disturbances at speed $c$.

  * **Negative (**$c^2 < 0$**):** If the sign flips (e.g., $u_{tt} = -|c^2| u_{xx}$), the equation changes type to **Elliptic** (becoming a Laplace equation in spacetime). Instead of propagating waves, the behavior becomes purely exponential growth/decay or steady-state.

  * **Complex (**$c \in \mathbb{C}$**):** Implies **dissipation** (the wave loses energy/attenuates as it travels) or **dispersion** (different frequency components travel at different speeds).

### 3. Laplace Equation

$$
0 = \nabla^2 u
$$

* **Physics & Applications:** * **Electrostatics:** Describes the electric potential in a region free of electric charge (Vacuum).

  * **Fluid Dynamics:** Potential flow of an incompressible, irrotational fluid.

  * **Heat Transfer:** Steady-state temperature distribution (thermal equilibrium) where temperature does not change with time.

  * **Gravitation:** Gravitational potential in empty space (vacuum).

  * **Elasticity:** Equilibrium shape of a stretched membrane (like a soap film) with no external load.

* **Characteristics:** No time dependence. The state is determined solely by the boundary conditions. It represents the "smoothest" possible interpolation between boundary values.

* **Nature:** Elliptic.

### 4. Poisson Equation (Laplace with Source)

$$
f(x, y, z) = \nabla^2 u
$$

* **Physics & Applications:** Describes potential fields driven by sources (mass, charge, heat).

  * **Electrostatics:** Electric potential $V$ in the presence of a charge density $\rho$ ($\nabla^2 V = -\rho/\epsilon_0$).

  * **Gravity:** Gravitational potential $\Phi$ in the presence of a mass density $\rho$ ($\nabla^2 \Phi = 4\pi G \rho$).

  * **Heat Transfer:** Steady-state temperature distribution with internal heat generation (e.g., electrical heating, chemical reaction).

  * **Fluid Dynamics:** Pressure field in incompressible flows (Pressure Poisson Equation) required to enforce continuity.

* **Nature:** Elliptic.

### 5. Schrödinger Equation

$$
i\hbar \frac{\partial \Psi}{\partial t} = -\left[ \frac{\hbar^2}{2m} \nabla^2 + V(x,t) \right] \Psi
$$

* **Physics:** Evolution of the Quantum Mechanical wavefunction.

## 3. Classification of PDEs

### (A) Order of PDEs

The order is defined by the **highest derivative** present in the equation. Most common physical PDEs are 2nd order.

### (B) Linearity

An equation is **Linear** if $u$ and its derivatives appear only to the first power. Otherwise, it is **Non-linear**.

* *Example of Non-linear:* Navier-Stokes Equation.

  $$
  \frac{\partial \mathbf{v}}{\partial t} + (\mathbf{v} \cdot \nabla) \mathbf{v} = -\nabla \left(\frac{p}{\rho}\right) + \nu \nabla^2 \mathbf{v}
  $$

  *(The convective term* $(\mathbf{v} \cdot \nabla) \mathbf{v}$ *makes it non-linear).*

  **Term-by-Term Breakdown:**

  * $\frac{\partial \mathbf{v}}{\partial t}$:  **Acceleration.** The rate of change of velocity with respect to time at a fixed point.

  * $(\mathbf{v} \cdot \nabla) \mathbf{v}$: **Convective Acceleration.** The rate of change of velocity due to the movement of fluid particles to a new location. This non-linear term is responsible for turbulence.

  * $-\nabla \left(\frac{p}{\rho}\right)$: **Pressure Gradient.** The force density driving the fluid flow from high pressure to low pressure.

  * $\nu \nabla^2 \mathbf{v}$: **Viscous Diffusion.** The diffusion of momentum due to internal fluid friction (kinematic viscosity $\nu$), which acts to smooth out velocity gradients.

* *Why is linearity important?* Linear equations allow for the **superposition of solutions**.

### (C) Fundamental Types

PDES are classified into three categories based on their behavior, analogous to conic sections ($Ax^2 + Bxy + Cy^2 + ...$).

Consider a general linear 2nd-order PDE in 2 variables:

$$
A u_{xx} + 2B u_{xy} + C u_{yy} + (\text{lower order terms}) = 0
$$

where $A, B, C \in \mathbb{R}$.

The discriminant is $D = B^2 - AC$.

| **Type** | **Discriminant** | **Prototype Example** | **Physical Phenomenon** | 
 | ----- | ----- | ----- | ----- | 
| **Hyperbolic** | $B^2 - AC > 0$ | $u_{tt} - c^2 u_{xx} = 0$ (Wave Eq) | Propagation. Disturbances propagate at finite speed. | 
| **Parabolic** | $B^2 - AC = 0$ | $u_t - k u_{xx} = 0$ (Heat Eq) | Diffusion / Smoothing processes. Disturbances propagate infinitely fast but decay rapidly. | 
| **Elliptic** | $B^2 - AC < 0$ | $u_{xx} + u_{yy} = 0$ (Laplace Eq) | Steady state / Equilibrium. Solution determined by boundary conditions (B.C.). | 

**General Form (Higher Dimensions):**

$$
L u = \sum_{i,j} a^{ij} \partial_i \partial_j u + \text{lower order terms} = 0
$$

Let $\bar{A} = [a^{ij}]$ be the coefficient matrix. Look at its eigenvalues:

* **Elliptic:** All eigenvalues have the same sign (all +ve or all -ve).

* **Hyperbolic:** One eigenvalue has a different sign than the rest (e.g., one +ve, rest -ve).

* **Parabolic:** One eigenvalue is zero, the rest have the same sign.

> **Note:** PDEs have infinite solutions. A specific problem is defined by the PDE + Boundary Conditions (B.C.) / Initial Data.

## 4. Numerical Solution of PDEs (with spectral methods)

**Goal:** Compute solutions of partial differential equations.

* *Hyperbolic/Parabolic:* Evolution equations (e.g., $\dot{u}(x,t) = u''(x,t)$).

* *Elliptic:* Boundary value problems (e.g., $u''(x) = 0$).

* *Eigenvalue problems:* $u''(x) = \lambda u(x)$ (e.g., modes of an oscillating string, Quasi-normal modes of Black Holes).

**Fundamental Idea of Spectral Methods:**
Expand the solution $u(x)$ in terms of **basis functions**:

$$
u(x) \approx \sum_{k=0}^{N} a_k \phi_k(x)
$$

Common basis functions: Fourier Series, Chebyshev Series.

### Introductory Example

**Problem:** Solve the ODE

$$
(E1) \quad u''(x) - (x^6 + 3x^2)u(x) = 0
$$

**Domain:** $x \in [-1, 1]$
**Boundary Conditions (B.C.):** $u(-1) = 1, \quad u(+1) = 1$.

**Trial Solution (Taylor Series):**
We start by expanding $u(x)$ as a polynomial series, which is fine since $|x| \leq 1$. To automatically satisfy the B.C., we can structure it as:

$$
u^{(N)} \approx 1 + (1-x^2)(a_0 + a_1 x + a_2 x^2 + \dots)
$$

For $N=4$ (order expansion):

$$
u^{(4)} = 1 + (1-x^2)(a_0 + a_1 x + a_2 x^2)
$$

**Consequences of Expanding:**

1. Can evaluate $u^{(N)}(x)$ **everywhere** (NOT just at grid points).

2. Can compute derivatives **EXACTLY**:

   $$
   u^{(4)'} = -2x(a_0 + a_1 x + a_2 x^2) + (1-x^2)(a_1 + 2a_2 x)
   $$

3. Can compute the **Residual** for Eqn (E1):

   $$
   R_N(x) = u^{(N)''}(x) - (x^6 + 3x^2)u^{(N)}(x)
   $$

   Because $u^{(N)}$ is an approximation, $R_N(x)$ will **NOT** be identically zero.

**Solving:**
To find $a_0, a_1, a_2$, we need equations.

Method: **Collocation**. We force the residual to vanish at specific grid points.
For 3 unknowns ($a_0, a_1, a_2$), pick 3 points (e.g., $x = -0.5, 0, 0.5$).

$$
R(-0.5) = 0
$$

$$
R(0) = 0
$$

$$
R(0.5) = 0
$$

Solve for $a_0, a_1, a_2$. Substitute back into $u^{(N)}$. **DONE.**

## 5. Spectral Methods

1. **Fourier Series:** Best for **Periodic** domains.

2. **Chebyshev Series:** Best for **Non-periodic** domains.

These methods allow us to:

* Expand solutions/functions.

* Calculate derivatives (Derivative Matrix).

* Use expansion for interpolation.

### 5.1 Fourier Series

Definition:

$$
u(x) = \frac{a_0}{2} + \sum_{k=1}^{\infty} a_k \cos(kx) + \sum_{k=1}^{\infty} b_k \sin(kx)
$$

Using Euler's formula ($e^{ikx} = \cos kx + i \sin kx$):

$$
u(x) = \sum_{k=-\infty}^{\infty} \tilde{c}_k e^{ikx}
$$

Relationship: $\tilde{c}_k = \frac{1}{2}(a_k - i b_k)$ and $\tilde{c}_{-k} = \tilde{c}_k^*$ (if $u(x)$ is real).

**Computing Coefficients:**
Using the orthogonality of Fourier basis functions:

$$
\int_{0}^{2\pi} e^{-imx} e^{ikx} dx = 2\pi \delta_{mk}
$$

We get:

$$
\tilde{c}_k = \frac{1}{2\pi} \int_{0}^{2\pi} u(x) e^{-ikx} dx
$$

#### Convergence of Fourier Series (Behavior of $c_n$ for large $n$)

Evaluate via Integration by Parts (IBP):

$$
2\pi \tilde{c}_n = \int_{0}^{2\pi} u(x) e^{-inx} dx
$$

Let $u = u(x)$ and $dv = e^{-inx}dx \implies v = \frac{i}{n}e^{-inx}$.

$$
= \left[ u(x) \frac{i}{n} e^{-inx} \right]_0^{2\pi} - \frac{i}{n} \int_{0}^{2\pi} u'(x) e^{-inx} dx
$$

**Case A:** $u(x)$ **is Periodic**
The boundary term $\frac{i}{n} (u(2\pi) - u(0))$ vanishes.
We can repeat IBP $k$ times.

$$
\tilde{c}_n = \left(\frac{-i}{n}\right)^k \int_{0}^{2\pi} u^{(k)}(x) e^{-inx} dx
$$

If $u(x) \in C^{\infty}$ (smooth and periodic):

* $\tilde{c}_n$ decays faster than any power of $(1/n)$.

* This is called **Exponential Convergence** (or Spectral Accuracy).

* Truncated Fourier series is very accurate.

**Case B:** $u(x)$ **is NOT Periodic**
The boundary terms do not vanish at the first integration step.

$$
\tilde{c}_n = \frac{1}{2\pi} \left[ \frac{u(x) e^{-inx}}{-in} \right]_0^{2\pi} - \frac{1}{2\pi} \int_{0}^{2\pi} u'(x) \frac{e^{-inx}}{-in} dx
$$

$$
= \frac{i}{2\pi n} (u(2\pi) - u(0)) + O(1/n^2)
$$

Since $u(2\pi) \neq u(0)$, the first term is non-zero.

$$
\tilde{c}_n \sim \frac{1}{n}
$$

Decay is **Algebraic** (slow). This corresponds to a "jump" in the function values at the boundaries (effectively a discontinuity when wrapped around).

**Case B (Special):** $u(x)$ **is NOT periodic, but its k-th derivative** $u^{(k)}(x)$ **is periodic**
Even if a higher derivative is periodic (e.g., $u(x) = x$ on $[0, 2\pi]$, where $u'$ is periodic), the convergence rate is determined by the **lowest order** term that violates periodicity or smoothness.
Since $u(x)$ itself is not periodic, the boundary term from the first integration by parts does not vanish.
Therefore, the decay remains:

$$
\tilde{c}_n \sim \frac{1}{n}
$$

**Case C:** $u(x)$ **is Periodic and** $C^k$ **(k-th derivative is smooth, but k+1 is discontinuous)**
Suppose $u(x)$ and its first $k$ derivatives are continuous and periodic ($u \in C^k_{per}$), but the $(k+1)$-th derivative $u^{(k+1)}(x)$ has discontinuities (jumps).

We perform Integration by Parts (IBP) $k+1$ times. Since $u, u', \dots, u^{(k)}$ are periodic, the boundary terms vanish at every step:

$$
\tilde{c}_n = \frac{1}{2\pi} \int_{0}^{2\pi} u(x) e^{-inx} dx
$$

$$
= \frac{1}{2\pi} \left( \frac{1}{in} \right) \int_{0}^{2\pi} u'(x) e^{-inx} dx \quad \text{(1st IBP)}
$$

$$
\vdots
$$

$$
= \frac{1}{2\pi} \left( \frac{1}{in} \right)^{k+1} \int_{0}^{2\pi} u^{(k+1)}(x) e^{-inx} dx \quad \text{((k+1)-th IBP)}
$$

The remaining integral is simply the Fourier coefficient of the $(k+1)$-th derivative, let's call it $\tilde{v}_n$:

$$
\tilde{c}_n = \frac{1}{(in)^{k+1}} \tilde{v}_n \quad \text{where } v(x) = u^{(k+1)}(x)
$$

Since $u^{(k+1)}(x)$ is discontinuous, its Fourier coefficients $\tilde{v}_n$ behave like **Case B** (decay as $1/n$).
Therefore, the total decay is:

$$
\tilde{c}_n \sim \frac{1}{n^{k+1}} \cdot \frac{1}{n} = \frac{1}{n^{k+2}}
$$

This leads to algebraic (slow) decay rather than exponential.

**Case D: Discontinuities**
**Gibbs Phenomenon:** $O(1)$ errors at discontinuities, constant overshoot independent of $N$.

> **Summary:** If $u$ is smooth but **not periodic**, Fourier is poor. Use **Chebyshev polynomials**. (Go to Pg 10 of notes).

#### 5.1.1 Numerical Integration (Gaussian Quadrature)

To compute coefficients numerically, we use quadrature.
We approximate the integral of a function $f(x)$ as a weighted sum:

$$
\int_{-1}^{1} f(x) w(x) dx \approx \sum_{j=0}^{N} w_j f(x_j)
$$

where $x_j$ are nodes and $w_j$ are weights.

**Gaussian Quadrature Theorem:**
For a given weight function, there exist nodes $x_j$ and weights $w_j$ such that the summation is **EXACT** for any polynomial $f$ of degree $\le 2N - 1$.

#### 5.1.2 Computing Fourier Series in Practice (DFT)

For periodic functions, we use the discrete version.
Let $N$ be even. Define a uniform grid on $[0, 2\pi]$:

$$
x_j = \frac{2\pi j}{N}, \quad j = 0, 1, \dots, N-1
$$

**Discrete Coefficients:**

$$
\tilde{c}_k \approx \frac{1}{2\pi} \frac{2\pi}{N} \sum_{j=0}^{N-1} u(x_j) e^{-ik x_j}
$$

$$
\tilde{c}_k = \frac{1}{N} \sum_{j=0}^{N-1} u_j e^{-ik x_j}
$$

This is the **Discrete Fourier Transform (DFT)**. Can be computed via **FFT** in $O(N \log N)$.

* **Note:** A single application of the Fast Fourier Transform (FFT) algorithm computes **ALL** coefficients $\{\tilde{c}_k\}_{k=0}^{N-1}$ simultaneously. You do *not* run a separate summation for each $k$.

* **Complexity:** The entire set of coefficients is obtained in $O(N \log N)$ operations, whereas direct summation for each $k$ would take $O(N^2)$.

**Inverse Transform:**

$$
u_j = \text{Re} \sum_{k=-N/2}^{N/2-1} \tilde{c}_k e^{ik x_j}
$$

##### Derivatives with Fourier Series

$$
u(x) = \sum \tilde{c}_k e^{ikx}
$$

$$
\frac{du}{dx} = \sum (ik) \tilde{c}_k e^{ikx} = \sum \tilde{c}'_k e^{ikx}
$$

where $\tilde{c}'_k = ik \tilde{c}_k$.

Process:

1. $u_j \xrightarrow{FFT} \tilde{c}_k$

2. Multiply by $ik$.

3. $\tilde{c}'_k \xrightarrow{IFFT} u'_j$

$$

$$

### 5.2 Non-Periodic Functions: Chebyshev Series

For non-periodic functions $u(x)$ on $x \in [-1, 1]$. (For arbitrary $[a, b]$, rescale $x \to \frac{x-b}{a}$).

**Chebyshev Polynomials:**

$$
T_k(x) = \cos(k \arccos x)
$$

Let $\theta = \arccos x$.

$$
T_0 = 1
$$

$$
T_1 = x
$$

$$
T_2 = 2x^2 - 1
$$

**Recurrence Relation:**

$$
T_{n+1}(x) = 2x T_n(x) - T_{n-1}(x), \quad n \ge 1
$$

**Orthogonality:**
Orthogonal with respect to weight function $\rho(x) = (1-x^2)^{-1/2}$.

$$
\int_{-1}^{1} T_m(x) T_n(x) \frac{dx}{\sqrt{1-x^2}} = \frac{\pi}{2} c_m \delta_{mn}
$$

where $c_0 = 2$, $c_m = 1$ for $m \neq 0$.

**Expansion:**

$$
u(x) = \sum_{k=0}^{\infty} \tilde{u}_k T_k(x)
$$

$$
\tilde{u}_k = \frac{2}{\pi c_k} \int_{-1}^{1} u(x) T_k(x) \rho(x) dx
$$

#### 5.2.1 Chebyshev Grids

1. **Chebyshev-Gauss Points:** Roots of $T_N(x)$.

   $$
   x_j = \cos\left( \frac{2j+1}{2N} \pi \right)
   $$

   (Does not include boundaries $\pm 1$).

2. **Gauss-Lobatto Points (Preferred for B.C.s):** Extrema of $T_N(x)$ plus endpoints.

   $$
   x_j = \cos\left( \frac{\pi j}{N} \right), \quad j = 0, 1, \dots, N
   $$

   These are the projection of equidistant points on a circle onto the x-axis.

#### 5.2.2 Discrete Quadrature (Gauss-Lobatto)

$$
\int_{-1}^{1} f(x) \rho(x) dx \approx \sum_{j=0}^{N} w_j f(x_j)
$$

This is exact for polynomials of degree $\le 2N - 1$.

#### 5.2.3 Discrete Orthogonality

$$
\sum_{j=0}^{N} \frac{1}{c_j} T_m(x_j) T_n(x_j) = \frac{N}{2} \bar{c}_n \delta_{mn}
$$

where $\bar{c}_N = 2$, $\bar{c}_n = 1$ otherwise.

#### 5.2.4 Derivatives in Chebyshev Space

We expand $u$ and $u'$ in Chebyshev series:

$$
u = \sum \tilde{u}_k T_k(x)
$$

$$
u' = \sum \tilde{u}'_k T_k(x)
$$

**Recurrence Relation for Derivatives:**
Using property: $\frac{T'_{k+1}}{k+1} - \frac{T'_{k-1}}{k-1} = 2 T_k$.
This leads to a backward recurrence relation to find coefficients of derivative $\tilde{u}'$:

$$
c_k \tilde{u}'_k = \tilde{u}'_{k+2} + 2(k+1)\tilde{u}_{k+1}
$$

*(Note:* $\tilde{u}'_N = 0, \tilde{u}'_{N+1} = 0$ *for truncated series)*.

**Algorithm:**

1. Set $\tilde{u}'_N = 0$, $\tilde{u}'_{N-1} = 0$.

2. Loop backwards from $k = N-2$ down to $0$.

3. Compute $\tilde{u}'_k$ using the formula.

**Matrix Form:**
This operation is linear and can be represented by an upper-triangular matrix $\mathbf{D}$.

$$
\mathbf{\tilde{u}'} = \mathbf{D} \mathbf{\tilde{u}}
$$

Example structure ($N=6$):

$$
\tilde{u}'_k = \frac{1}{c_k} (\tilde{u}'_{k+2} + 2(k+1)\tilde{u}_{k+1})
$$

(Creates a matrix with non-zero entries on super-diagonals).

$$

## 6. Operators in Matrix Form

Differentiation and transforms are linear operations. We can represent the relationship between physical space values $\mathbf{u}$ and spectral coefficients $\mathbf{\tilde{c}}$ using matrix algebra.

$$
\mathbf{\tilde{c}} = \mathbf{S} \mathbf{u}
$$

$$
\mathbf{u} = \mathbf{S}^{-1} \mathbf{\tilde{c}}
$$

**Derivative Calculation:**

$$
\mathbf{u} \to \mathbf{\tilde{c}} \to \mathbf{\tilde{c}}' \to \mathbf{u}'
$$

$$
\mathbf{u}' = \mathbf{S}^{-1} \mathbf{D}_{spec} \mathbf{S} \mathbf{u}
$$

where $\mathbf{D}_{spec} = \text{diag}(0, i, \dots, ik, \dots)$.

This defines a **Differentiation Matrix** in real space:

$$
\frac{du}{dx}(x_i) \approx \sum_{j} D_{ij} u_j
$$

This yields derivatives at the full accuracy of the truncated Fourier series.

### 6.1 Derivation of operators: Fourier Series

Consider we keep $N$ terms on a grid $x \in [0, 2\pi)$.
The grid points are $x_j = \frac{2\pi j}{N}$ for $j = 0, 1, \dots, N-1$.
Indices for spectral coefficients $k$ are centered around zero:

* **Even N:** $k \in \{-N/2, \dots, N/2 - 1\}$

* **Odd N:** $k \in \{-(N-1)/2, \dots, (N-1)/2\}$

#### 6.1.1 The Forward Transform Matrix ($S$)

The coefficients $\tilde{c}_k$ are computed via the Discrete Fourier Transform (DFT):

$$
\tilde{c}_k = \frac{1}{N} \sum_{j=0}^{N-1} u_j e^{-ik x_j}
$$

Writing this as a matrix-vector multiplication $\mathbf{\tilde{c}} = \mathbf{S} \mathbf{u}$:
The element $S_{kj}$ in the $k$-th row and $j$-th column is:

$$
S_{kj} = \frac{1}{N} e^{-ik x_j} = \frac{1}{N} e^{-i \frac{2\pi k j}{N}}
$$

**Explicit Matrices:**

* **Case** $N=2$ ($k \in \{-1, 0\}$):

  $$
  \mathbf{S} = \frac{1}{2}
  \begin{pmatrix}
  1 & -1 \\
  1 & 1
  \end{pmatrix}
  $$

* **Case** $N=3$ ($k \in \{-1, 0, 1\}$):
  Let $\omega = e^{-i 2\pi/3}$.

  $$
  \mathbf{S} = \frac{1}{3}
  \begin{pmatrix}
  1 & \omega^2 & \omega \\
  1 & 1 & 1 \\
  1 & \omega & \omega^2
  \end{pmatrix}
  $$

* **Case** $N=4$ ($k \in \{-2, -1, 0, 1\}$):

  $$
  \mathbf{S} = \frac{1}{4}
  \begin{pmatrix}
  1 & -1 & 1 & -1 \\
  1 & i & -1 & -i \\
  1 & 1 & 1 & 1 \\
  1 & -i & -1 & i
  \end{pmatrix}
  $$

* **Case** $N=5$ ($k \in \{-2, \dots, 2\}$):
  Let $\omega = e^{-i 2\pi/5}$.

  $$
  \mathbf{S} = \frac{1}{5}
  \begin{pmatrix}
  1 & \omega^3 & \omega^6 & \omega^9 & \omega^{12} \\
  1 & \omega^4 & \omega^8 & \omega^{12} & \omega^{16} \\
  1 & 1 & 1 & 1 & 1 \\
  1 & \omega & \omega^2 & \omega^3 & \omega^4 \\
  1 & \omega^2 & \omega^4 & \omega^6 & \omega^8
  \end{pmatrix}
  $$

* **Case** $N=6$ ($k \in \{-3, \dots, 2\}$):
  Let $\omega = e^{-i\pi/3}$.

  $$
  \mathbf{S} = \frac{1}{6}
  \begin{pmatrix}
  1 & -1 & 1 & -1 & 1 & -1 \\
  1 & \omega^{-2} & \omega^{-4} & 1 & \omega^{-2} & \omega^{-4} \\

### 6.2 Derivation of operators: Chebyshev series

Chebyshev spectral methods rely on mapping the non-periodic domain $x \in [-1, 1]$ to the periodic domain $\theta \in [0, \pi]$ via the transformation $x = \cos \theta$. We use the **Gauss-Lobatto** grid (which includes endpoints) to allow for boundary conditions.

Consider $N$ grid points (corresponding to polynomial degree $M = N-1$).
The grid points are $x_j = \cos\left(\frac{\pi j}{N-1}\right)$ for $j = 0, \dots, N-1$.
The Chebyshev weights are defined as $c_j = 2$ for $j=0, N-1$ and $c_j = 1$ otherwise.

**Derivation of Weights:**
These weights arise from applying the **Trapezoidal Rule** to the integral in the mapped variable $\theta$.

1. Start with the weighted integral: $I = \int_{-1}^1 f(x) \frac{dx}{\sqrt{1-x^2}}$.

2. Substitute $x = \cos \theta$. Then $dx = -\sin \theta d\theta$ and $\sqrt{1-x^2} = \sin \theta$. The integral simplifies to:

   $$
   I = \int_0^\pi f(\cos \theta) d\theta
   $$

3. Discretize $\theta$ using $N$ equally spaced points: $\theta_j = \frac{\pi j}{N-1}$ for $j=0, \dots, N-1$. The step size is $\Delta \theta = \frac{\pi}{N-1}$.

4. Apply the Trapezoidal Rule:

   $$
   I \approx \Delta \theta \left[ \frac{1}{2}f(\cos \theta_0) + \sum_{j=1}^{N-2} f(\cos \theta_j) + \frac{1}{2}f(\cos \theta_{N-1}) \right]
   $$

5. Comparing this to the summation form $\frac{\pi}{N-1} \sum_{j=0}^{N-1} \frac{1}{c_j} f(x_j)$, we identify the coefficients:

   * Endpoints ($j=0, N-1$): Coefficient is $1/2 \implies 1/c_j = 1/2 \implies c_j = 2$.

   * Interior ($j=1 \dots N-2$): Coefficient is $1 \implies 1/c_j = 1 \implies c_j = 1$.

#### 6.2.1 The Forward Transform Matrix ($S$)

To derive the matrix form, we start from the continuous orthogonality of Chebyshev polynomials:

$$
\int_{-1}^{1} T_k(x) T_m(x) \frac{dx}{\sqrt{1-x^2}} = \frac{\pi}{2} \bar{c}_k \delta_{km}
$$

where $\bar{c}_k$ are standardization constants ($\bar{c}_0=2, \bar{c}_k=1$ for $k>0$).
The spectral coefficients $\tilde{c}_k$ for a function $u(x)$ are obtained by projecting $u(x)$ onto the basis $T_k(x)$:

$$
\tilde{c}_k = \frac{2}{\pi \bar{c}_k} \int_{-1}^{1} u(x) T_k(x) \frac{dx}{\sqrt{1-x^2}}
$$

We approximate this integral using **Gauss-Lobatto Quadrature**. For an integral weighted by $1/\sqrt{1-x^2}$, the discrete approximation is:

$$
\int_{-1}^1 f(x) \frac{dx}{\sqrt{1-x^2}} \approx \frac{\pi}{N-1} \sum_{j=0}^{N-1} \frac{1}{c_j} f(x_j)
$$

Substituting $f(x) = u(x) T_k(x)$ into the projection integral:

$$
\tilde{c}_k \approx \frac{2}{\pi \bar{c}_k} \left( \frac{\pi}{N-1} \sum_{j=0}^{N-1} \frac{1}{c_j} u(x_j) T_k(x_j) \right)
$$

Using the grid definition $T_k(x_j) = \cos\left(\frac{\pi k j}{N-1}\right)$ and simplifying, we arrive at the Discrete Chebyshev Transform. Note that for the finite discrete transform, the coefficient $c_k$ in the denominator matches the grid weight definition $c_j$ (it becomes 2 at $k=N-1$ as well due to aliasing on the grid).

$$
\tilde{c}_k = \frac{2}{(N-1) c_k} \sum_{j=0}^{N-1} \frac{1}{c_j} u_j \cos\left(\frac{\pi k j}{N-1}\right)
$$

Writing this as $\mathbf{\tilde{c}} = \mathbf{S} \mathbf{u}$:
The element $S_{kj}$ is:

$$
S_{kj} = \frac{2}{(N-1) c_k c_j} \cos\left(\frac{\pi k j}{N-1}\right)
$$

#### 6.2.3 The Spectral Derivative Matrix ($D_{spec}$)

The derivative of a Chebyshev expansion is computed via the recurrence relation: 

$$2k \tilde{c}_k = c_{k-1} \tilde{c}'_{k-1} - \tilde{c}'_{k+1}$$

By shifting the index ($k \to k+1$) and accounting for the boundary case $k=0$ with the standard weight $c_k$ ($c_0=2$, $c_{k>0}=1$), we can rearrange this to solve for $\tilde{c}'_k$:

$$c_k \tilde{c}'_k = 2(k+1)\tilde{c}_{k+1} + \tilde{c}'_{k+2}$$

To find an explicit formula for $\tilde{c}'_k$, we can recursively expand the $\tilde{c}'_{k+2}$ term on the right-hand side. Since $c_{k+2} = 1$ (as $k+2 \ge 2$), substituting the relation for $k+2$ yields:

$$c_k \tilde{c}'_k = 2(k+1)\tilde{c}_{k+1} + \left( 2(k+3)\tilde{c}_{k+3} + \tilde{c}'_{k+4} \right)$$

Continuing this telescoping expansion until we reach the truncation limit $N$ (where $\tilde{c}'_N = \tilde{c}'_{N+1} = 0$), we get a sum of alternating terms:

$$c_k \tilde{c}'_k = 2(k+1)\tilde{c}_{k+1} + 2(k+3)\tilde{c}_{k+3} + 2(k+5)\tilde{c}_{k+5} + \dots$$

Dividing by $c_k$, we arrive at the explicit formula where the coefficients of the derivative $\tilde{c}'$ are linear combinations of the coefficients $\tilde{c}$:

$$
\tilde{c}'_k = \frac{2}{c_k} \sum_{\substack{p=k+1 \\ p+k \text{ odd}}}^{N-1} p \, \tilde{c}_p
$$

Since $\tilde{c}_k$ depents on $\tilde{c}_\ell$ with $\ell\geq k+1$, the derivative matrix is strictly **upper triangular**.

**Explicit Matrices:**

* **Case** $N=2$:

  $$

  0 & 0
  \end{pmatrix}
  $$

* **Case** $N=3$:

  $$

  \begin{pmatrix}
  0 & 1 & 0 \\
  0 & 0 & 4 \\
  0 & 0 & 0
  \end{pmatrix}
  $$

* **Case** $N=4$:

  $$
  \mathbf{D}_{spec} =
  \begin{pmatrix}
  0 & 1 & 0 & 3 \\
  0 & 0 & 4 & 0 \\

  0 & 0 & 0 & 0
  \end{pmatrix}
  $$

* **Case** $N=5$:

  \begin{pmatrix}

* **Case** $N=6$:

To see how the formula constructs this matrix, evaluate the explicit sum $\tilde{c}'_k = \frac{2}{c_k} \sum_{\substack{p=k+1 \\ p+k \text{ odd}}}^{5} p \, \tilde{c}_p$ for each row $k=0, \dots, 5$, noting $c_0=2$ and $c_{k>0}=1$:

- Row 1 ($k=1$): $\tilde{c}'_1 = \frac{2}{1} (2\tilde{c}_2 + 4\tilde{c}_4) = \mathbf{4}\tilde{c}_2 + \mathbf{8}\tilde{c}_4$

- Row 2 ($k=2$): $\tilde{c}'_2 = \frac{2}{1} (3\tilde{c}_3 + 5\tilde{c}_5) = \mathbf{6}\tilde{c}_3 + \mathbf{10}\tilde{c}_5$

#### 6.2.4 The Physical Differentiation Matrix ($D$)

The operator acting on physical grid values is constructed using analytic derivatives of the Lagrange polynomials at the Gauss-Lobatto points.

**Explicit Derivation:**

1. **Lagrange Basis:**
   We represent the interpolant as $u(x) \approx \sum_{j=0}^N u_j \ell_j(x)$, where $\ell_j(x)$ are Lagrange polynomials defined by:

   $$
   \ell_j(x) = \frac{\omega(x)}{\omega'(x_j)(x-x_j)}
   $$

   where $\omega(x) = \prod_{k=0}^N (x-x_k)$ is the node polynomial.

2. **Node Polynomial for Gauss-Lobatto:**
   The nodes $x_j = \cos(\pi j/N)$ are the roots of $\omega(x) = (1-x^2)T'_N(x)$.

3. **Derivative of Basis Functions (**$D_{ij} = \ell_j'(x_i)$**):**
   Differentiating the expression for $\ell_j(x)$:

   $$
   \ell_j'(x) = \frac{\omega'(x)(x-x_j) - \omega(x)}{\omega'(x_j)(x-x_j)^2}
   $$

   * **Case** $i \neq j$:
     Evaluating at $x = x_i$ (where $\omega(x_i) = 0$):

     $$
     D_{ij} = \ell_j'(x_i) = \frac{\omega'(x_i)(x_i-x_j)}{\omega'(x_j)(x_i-x_j)^2} = \frac{\omega'(x_i)}{\omega'(x_j)} \frac{1}{x_i - x_j}
     $$

     Using properties of Chebyshev polynomials, one can show $\omega'(x_j) = (-1)^{j+1} N^2 c_j$ (where $c_j=2$ at endpoints, $1$ otherwise).
     Substituting this yields:

     $$
     D_{ij} = \frac{c_i}{c_j} \frac{(-1)^{i+j}}{x_i - x_j}
     $$

   * **Case** $i = j$:
     Using L'Hôpital's rule on the derivative expression or the property $\sum_{k} D_{ik} = 0$ (derivative of a constant is 0):

     $$
     D_{jj} = -\sum_{k \neq j} D_{jk}
     $$

     Analytically, this evaluates to:

     $$
     D_{jj} = -\frac{x_j}{2(1-x_j^2)} \quad (j \neq 0, N)
     $$

     $$
     D_{00} = \frac{2N^2+1}{6}, \quad D_{NN} = -\frac{2N^2+1}{6}
     $$

**Explicit Matrices:**

* **Case** $N=2$:

  0.5 & -0.5
  \end{pmatrix}

  $$

  \end{pmatrix}
  $$

  \frac{1}{2}(1+\sqrt{2}) & -\frac{\sqrt{2}}{2} & -\sqrt{2} & \frac{\sqrt{2}}{2} & \frac{1}{2}(1-\sqrt{2}) \\

   * Plot: $\log |\tilde{u}_k|$ vs $\log k$ (Log-Log).

   * Result: Straight line. Slope $-p$.

   * Implies: $\tilde{u}_k \sim k^{-p}$. Function has $p$-th derivative discontinuity or is non-periodic (for Fourier).

* **Super-geometric:** Decays faster than exponential (e.g., $e^{-k^2}$).

* **Sub-geometric:** Slower than exponential.

**Practical Use:** Since $u(x)$ is often only known numerically, the decay rate of $\tilde{u}_k$ acts as a diagnostic tool to determine smoothness or resolution issues.

## 8. Solving Elliptic PDEs 

**Example:**

$$
u'' - (x^6 + 3x^2)u = 0
$$

$$
u(1) = u(-1) = 1
$$

**Step 1: Spectral Expansion**
Write $u^{(N)}(x) = \sum_{k=0}^{N} \tilde{u}_k T_k(x)$.
We can parameterize unknowns using spectral coefficients $\{\tilde{u}_k\}$ OR grid values $\{u_j = u(x_j)\}$.

**Step** 2: Substitute **into PDE**

$$
R(x) = u^{(N)''}(x) - (x^6 + 3x^2)u^{(N)}(x) \approx 0
$$

**Step 3: Collocation**
Enforce $R(x_i) = 0$ at grid points $i = 0, \dots, N$.
In Matrix form using grid values:

$$
\sum_{j} D^2_{ij} u_j - F(x_i) u_i = 0
$$

$$
(\mathbf{D}^2 - \mathbf{F}) \mathbf{u} = 0
$$

where $\mathbf{D}^2$ is the second derivative matrix and $\mathbf{F}$ is diagonal matrix of the function $(x^6+3x^2)$.

**Step 4: Boundary Conditions**
Currently, $(\mathbf{D}^2 - \mathbf{F})\mathbf{u} = 0$ gives the trivial solution $u=0$.
We must impose B.C.s: $u_0 = 1$ and $u_N = 1$.
Modify the matrix system:

1. Replace Row 0 with $[1, 0, \dots, 0]$ and RHS with $1$.

2. Replace Row N with $[0, \dots, 0, 1]$ and RHS with $1$.

$$
\begin{pmatrix} 1 & 0 & \dots & 0 \\ \dots & \dots & \dots & \dots \\ 0 & \dots & 0 & 1 \end{pmatrix}
\begin{pmatrix} u_0 \\ \dots \\ u_N \end{pmatrix}
=
\begin{pmatrix} 1 \\ 0 \\ 1 \end{pmatrix}
$$

**Step 5: Solve**

$$
\mathbf{u} = \mathbf{A}^{-1} \mathbf{b}
$$

**General** Linear **Form:**

$$
u'' + a(x)u' + b(x)u = c(x)
$$

Matrix system:

$$
(\mathbf{D}^2 + \mathbf{A}\mathbf{D} + \mathbf{B}) \mathbf{u} = \mathbf{c}
$$

## 9. Solving Parabolic PDEs (Time-Stepping)

**Example:** Heat Equation

$$
\frac{\partial u}{\partial t} = \frac{\partial^2 u}{\partial x^2} \quad \text{for } x \in [-1, 1]
$$

**Initial Condition:** $u(x,0) = e^{-x^2}$
**Boundary Conditions:** $u(-1, t) = u(1, t) = 0$

**Step 1: Method of Lines (Spatial Discretization)**
Instead of discretizing time immediately, we keep time continuous and discretize only space. We represent the solution as a vector of time-dependent grid values:

$$
\mathbf{u}(t) = \begin{pmatrix} u_0(t) \\ u_1(t) \\ \vdots \\ u_N(t) \end{pmatrix}
$$

Using the spectral differentiation matrix $\mathbf{D}^2$ (typically Chebyshev for bounded domains), we approximate the spatial derivative:

$$
\frac{\partial^2 u}{\partial x^2} \approx \mathbf{D}^2 \mathbf{u}(t)
$$

**Step 2: System of ODEs**
Substituting the approximation into the PDE yields a system of coupled Ordinary Differential Equations (ODEs):

$$
\frac{d\mathbf{u}}{dt} = \mathbf{D}^2 \mathbf{u}(t)
$$

**Step 3: Imposing Boundary Conditions**
The differentiation matrix $\mathbf{D}^2$ acts on all points, but the values at the boundaries $u_0(t)$ and $u_N(t)$ are fixed by the BCs (0 in this case).
We partition the system into boundary (B) and interior (I) points. Since $u_0, u_N$ are constant, their time derivatives are zero. We only solve for the interior:

$$
\frac{d\mathbf{u}_I}{dt} = (\mathbf{D}^2)_{II} \mathbf{u}_I + (\mathbf{D}^2)_{IB} \mathbf{u}_B
$$

Since $\mathbf{u}_B = 0$, this simplifies to:

$$
\frac{d\mathbf{u}_I}{dt} = \tilde{\mathbf{A}} \mathbf{u}_I
$$

where $\tilde{\mathbf{A}}$ is the $(N-1) \times (N-1)$ submatrix of $\mathbf{D}^2$ corresponding to interior nodes.

**Step 4: Time Integration**
We now solve this ODE system using a time-stepping scheme.

* **Option A: Explicit Euler (Simple but Unstable)**
  $\mathbf{u}^{n+1} = \mathbf{u}^n + \Delta t \tilde{\mathbf{A}} \mathbf{u}^n$
  *Problem:* For spectral methods, the eigenvalues of $\mathbf{D}^2$ scale as $O(N^4)$. Stability requires $\Delta t < C/N^4$, which is extremely restrictive.

* **Option B: Implicit Euler (Stable)**
  $\mathbf{u}^{n+1} = \mathbf{u}^n + \Delta t \tilde{\mathbf{A}} \mathbf{u}^{n+1}$
  Rearranging gives a linear system to solve at every time step:

  $$
  (\mathbf{I} - \Delta t \tilde{\mathbf{A}}) \mathbf{u}^{n+1} = \mathbf{u}^n
  $$

  This is unconditionally stable, allowing much larger time steps.

* **Option C: Crank-Nicolson (Accurate & Stable)**
  A mix of explicit and implicit (Trapezoidal rule in time):

  $$
  (\mathbf{I} - \frac{\Delta t}{2} \tilde{\mathbf{A}}) \mathbf{u}^{n+1} = (\mathbf{I} + \frac{\Delta t}{2} \tilde{\mathbf{A}}) \mathbf{u}^n
  $$

## 10. Solving Hyperbolic PDEs (Wave Evolution)

**Example**: Wave Equation

$$\frac{\partial^2 u}{\partial t^2} = c^2 \frac{\partial^2 u}{\partial x^2} \quad \text{for } x \in [-1, 1]$$

**Initial Conditions**: $u(x,0) = f(x)$ (initial position), $u_t(x,0) = g(x)$ (initial velocity).
Boundary Conditions: $u(-1, t) = u(1, t) = 0$ (Fixed ends).

**Step 1: Method of Lines**
Using the spectral differentiation matrix $\mathbf{D}^2$ for the spatial derivative:

$$\frac{d^2\mathbf{u}}{dt^2} = c^2 \mathbf{D}^2 \mathbf{u}(t)$$

Unlike the heat equation, this is a second-order ODE in time.

**Step 2: Reduction to First-Order System**
Most standard time-stepping algorithms (like Runge-Kutta) require a system of the form $\frac{d\mathbf{y}}{dt} = F(\mathbf{y}, t)$. We convert our second-order system by introducing the velocity vector $\mathbf{v}(t) = \frac{d\mathbf{u}}{dt}$.

State vector: $\mathbf{y}(t) = \begin{pmatrix} \mathbf{u}(t) \\ \mathbf{v}(t) \end{pmatrix}$.

The system becomes:

$$\frac{d}{dt} \begin{pmatrix} \mathbf{u} \\ \mathbf{v} \end{pmatrix} = \begin{pmatrix} \mathbf{v} \\ c^2 \mathbf{D}^2 \mathbf{u} \end{pmatrix}$$

**Step 3: Boundary Conditions**
We partition $\mathbf{u}$ and $\mathbf{v}$ into interior ($I$) and boundary ($B$) components.
For fixed ends ($u(\pm 1)=0$):

$\mathbf{u}_B = 0$

$\mathbf{v}_B = \frac{d}{dt}\mathbf{u}_B = 0$

Thus, we only evolve the interior points:

$$\frac{d}{dt} \begin{pmatrix} \mathbf{u}_I \\ \mathbf{v}_I \end{pmatrix} = \begin{pmatrix} \mathbf{v}_I \\ c^2 \tilde{\mathbf{A}} \mathbf{u}_I \end{pmatrix}$$

where $\tilde{\mathbf{A}}$ is the interior submatrix of $\mathbf{D}^2$.

**Step 4: Time Integration & Stability (CFL Condition)**

- **Explicit Schemes** (e.g., RK4): Commonly used for hyperbolic problems, as implicit schemes can be computationally heavy and explicit schemes preserve wave behavior better without artificial numerical dissipation.

- **Stability Constraint**: The Courant-Friedrichs-Lewy (CFL) condition dictates stability. For hyperbolic PDEs, stability generally requires $\Delta t < C \cdot \Delta x_{min}$.

- **Chebyshev stiffness**: Since the grid points cluster at the boundaries, $\Delta x_{min} \sim O(N^{-2})$.
Therefore, the maximum stable time step scales as:

$$\Delta t \sim O(N^{-2})$$

(*Note*: This is restrictive, but much better than the $O(N^{-4})$ restriction for the Heat equation).

## 11. Recap: Numerical PDE Strategy

**1. Elliptic Equations (Boundary Value Problems):**

* Entire PDE becomes a matrix equation $\mathbf{L}\mathbf{u} = \mathbf{f}$.

* Replace derivatives with Derivative Matrices ($\mathbf{D}, \mathbf{D}^2$).

* Modify rows for Boundary Conditions.

* Solve linear system.

**2. Hyperbolic / Parabolic Equations (Time Evolution):**

* $u_t = \text{RHS}(u, u_x, \dots)$.

* Use Spectral method for spatial derivatives: $u_x \to \mathbf{D}\mathbf{u}$.

* **Result is a system of ODEs in time:** $\frac{d\mathbf{u}}{dt} = \mathbf{D}\mathbf{u}$.

  **From PDE to ODE (The Method of Lines)**
  From first principles, this transformation occurs because we discretize space while keeping time continuous:

  1. **Semi-Discrete Ansatz:** We approximate the solution $u(x,t)$ essentially as a vector of time-dependent coefficients (or grid values): $\mathbf{u}(t) = [u_0(t), u_1(t), \dots, u_N(t)]^T$.

  2. **Spatial Operator Replacement:** Any spatial derivative operator $\mathcal{L}$ (like $\partial_x$ or $\partial_{xx}$) acts linearly on the spatial basis functions. In the discrete basis, this linear operator is represented exactly by a matrix $\mathbf{D}$. Thus, $\mathcal{L} u(x,t)$ becomes the matrix-vector product $\mathbf{D}\mathbf{u}(t)$.

  3. **The Transformation:**

     $$
     \underbrace{\frac{\partial u}{\partial t}}_{\text{PDE}} = \underbrace{\mathcal{L} u}_{\text{Spatial Op}} \quad \xrightarrow{\text{Discretize Space}} \quad \underbrace{\frac{d \mathbf{u}}{dt}}_{\text{ODE}} = \underbrace{\mathbf{D} \mathbf{u}}_{\text{Matrix Algebra}}
     $$

  4. **Conclusion:** The partial derivative $\partial_t$ becomes a total derivative $d/dt$ because the spatial dependency is now "frozen" into the grid indices. We are left with a system of coupled Ordinary Differential Equations for the values $u_i(t)$.

* Use ODE integrators (Runge-Kutta, etc.) to advance in time.

**Workflow:**
Physical Space ($u_j$) $\xrightarrow{\mathbf{S}}$ Spectral Space ($\tilde{c}_k$) $\xrightarrow{\text{deriv}}$ Spectral Derivative ($\tilde{c}'_k$) $\xrightarrow{\mathbf{S}^{-1}}$ Physical Derivative ($u'_j$).
*Or simply:*

$$
\mathbf{u}' = \mathbf{D}_{phys} \mathbf{u}
$$
